
### 1. What is Probability?

Probability is the mathematical measure of how likely an event
is to occur. Its value ranges from 0 to 1, where 0 means an
impossible event and 1 means a certain event.

**Formula:**

Probability of an event = Number of favourable outcomes /
Total number of equally likely outcomes

For observed data, empirical probability is calculated as:

Empirical Probability = Number of times the event occurs /
Total number of observations

### 2. Key Probability Terminology

1. **Experiment:** An activity or process that produces an outcome.
   Example: Selecting a student from the dataset.

2. **Outcome:** A possible result of an experiment.
   Example: The selected student passes the exam.

3. **Sample Space:** The set of all possible outcomes.
   Example: Pass or Fail.

4. **Event:** A specific outcome or group of outcomes.
   Example: Selecting a student who passes.

5. **Favourable Outcomes:** Outcomes that satisfy the event.
   Example: Students whose result is Pass.

6. **Complementary Event:** The event that the original event
   does not occur. If A means Pass, its complement means Fail.

7. **Empirical Probability:** Probability estimated from observed
   data.

8. **Theoretical Probability:** Probability calculated using
   equally likely outcomes and mathematical reasoning.

### 3. Three Probability Examples from the Dataset

1. Probability of selecting a student who passes the final exam.
2. Probability of selecting a student who participates in
   group discussions.
3. Probability of selecting a student whose attendance is
   greater than 80%.


## Task 2: Types of Events

### 1. Empirical Probability

Empirical probability is calculated using the actual observations
available in the dataset.

Formula:

P(A) = Number of times event A occurs / Total observations

In this project, we calculate the empirical probability of
a randomly selected student passing the final examination.

### 2. Theoretical Probability

Theoretical probability is calculated using the possible
equally likely outcomes of an experiment.

Formula:

P(A) = Number of favourable outcomes / Total possible outcomes

For a fair coin toss, the possible outcomes are Heads and Tails.
Both outcomes are equally likely.

Therefore, the theoretical probability of getting Heads is:

P(Heads) = 1 / 2 = 0.5 = 50%


```python

# Task 2: Empirical and Theoretical Probability

# Empirical probability from our student dataset
total_students = len(df)
pass_count = (df["final_exam_pass"] == "Pass").sum()

empirical_probability = pass_count / total_students

print("EMPIRICAL PROBABILITY")
print("Event: A randomly selected student passes the exam")
print(f"Favourable outcomes = {pass_count}")
print(f"Total observations = {total_students}")
print(f"P(Pass) = {pass_count}/{total_students}")
print(f"P(Pass) = {empirical_probability:.4f}")
print(f"P(Pass) = {empirical_probability * 100:.2f}%")

# Theoretical probability for a fair coin toss
favourable_outcomes = 1  # Heads
total_outcomes = 2       # Heads or Tails

theoretical_probability = favourable_outcomes / total_outcomes

print("\nTHEORETICAL PROBABILITY")
print("Event: Getting Heads in a fair coin toss")
print(f"P(Heads) = {favourable_outcomes}/{total_outcomes}")
print(f"P(Heads) = {theoretical_probability:.2f}")
print(f"P(Heads) = {theoretical_probability * 100:.0f}%")
```


## Task 3: Random Variable and Probability Distribution

### 1. Definition of Random Variable

Let X be the number of students who pass the final examination
among 3 randomly selected students.

The possible values of X are 0, 1, 2, and 3.

We assume that each selected student has the same probability
of passing, based on the observed dataset, and that the selections
are independent for this probability model.

The estimated probability of passing is:

p = 145 / 200 = 0.725

The probability of failing is:

q = 1 - p = 0.275

Therefore, X follows a Binomial Probability Distribution:

X ~ Binomial(n = 3, p = 0.725)

### 2. Probability Distribution Formula

P(X = x) = C(n, x) × p^x × (1-p)^(n-x)

Where:
- n = 3 selected students
- x = Number of students who pass
- p = Probability of passing
- 1-p = Probability of failing
- C(n, x) = Number of combinations


```python

# Task 3: Probability Distribution for 3 Students

from math import comb

n = 3
p = (df["final_exam_pass"] == "Pass").mean()
q = 1 - p

x_values = [0, 1, 2, 3]

probabilities = [
    comb(n, x) * (p ** x) * (q ** (n - x))
    for x in x_values
]

distribution = pd.DataFrame({
    "X (Number of Passes)": x_values,
    "P(X = x)": probabilities
})

print("Probability Distribution: Number of Passes out of 3")
display(distribution.round(4))

print("Sum of probabilities:", round(sum(probabilities), 4))
```

```python

# Mean and Variance of the Binomial Random Variable

mean_x = n * p
variance_x = n * p * q
standard_deviation_x = variance_x ** 0.5

print(f"Probability of passing (p): {p:.4f}")
print(f"Probability of failing (q): {q:.4f}")

print(f"\nMean E(X) = n × p = {n} × {p:.4f} = {mean_x:.4f}")
print(
    f"Variance Var(X) = n × p × q "
    f"= {n} × {p:.4f} × {q:.4f} = {variance_x:.4f}"
)
print(f"Standard deviation = {standard_deviation_x:.4f}")
```


## Task 4: Venn Diagram in Probability

A Venn diagram is a graphical representation of sets and their
relationships.

For this project, we define two events:

- **Event A:** Students who study more than 10 hours per week.
- **Event B:** Students whose attendance is greater than 80%.

The intersection A ∩ B represents students who satisfy both
conditions.

The diagram also shows students who satisfy only one condition
and students who satisfy neither condition.


```python

import matplotlib.pyplot as plt
from matplotlib_venn import venn2

# Define the two events
event_A = df["study_hours"] > 10
event_B = df["attendance"] > 80

# Count each region
A_only = int((event_A & ~event_B).sum())
B_only = int((event_B & ~event_A).sum())
both = int((event_A & event_B).sum())
neither = int((~event_A & ~event_B).sum())

print("Venn Diagram Counts")
print("Study > 10 hours only:", A_only)
print("Attendance > 80% only:", B_only)
print("Both conditions:", both)
print("Neither condition:", neither)
print("Total students:", A_only + B_only + both + neither)

# Draw the Venn diagram
plt.figure(figsize=(8, 6))

venn2(
    subsets=(A_only, B_only, both),
    set_labels=(
        "Study > 10 hours/week",
        "Attendance > 80%"
    )
)

plt.title("Venn Diagram: Study Hours and Attendance")
plt.tight_layout()
plt.show()
```


## Task 5: Contingency Table and Probability

A contingency table is a table that shows the frequency
distribution of two categorical variables.

In this project, we study the relationship between:

1. Group Discussion: Yes or No
2. Final Exam Result: Pass or Fail

The table helps us compare exam outcomes among students
who participate in group discussions and those who do not.

We will also calculate joint probability and conditional
probability using the observed dataset.


```python

# Task 5: Contingency Table

contingency_table = pd.crosstab(
    df["group_discussion"],
    df["final_exam_pass"],
    margins=True
)

print("Contingency Table: Group Discussion vs Exam Result")
display(contingency_table)
```

```python

# Probability calculations

total = len(df)

# 1. Joint probability: Discussion = Yes AND Pass
joint_count = (
    (df["group_discussion"] == "Yes") &
    (df["final_exam_pass"] == "Pass")
).sum()

joint_probability = joint_count / total

# 2. Conditional probability: Pass given Discussion = Yes
discussion_yes_count = (df["group_discussion"] == "Yes").sum()

pass_given_discussion = joint_count / discussion_yes_count

# 3. Probability of passing without group discussion
discussion_no_count = (df["group_discussion"] == "No").sum()

pass_without_discussion_count = (
    (df["group_discussion"] == "No") &
    (df["final_exam_pass"] == "Pass")
).sum()

pass_given_no_discussion = (
    pass_without_discussion_count / discussion_no_count
)

print("1. Joint Probability: P(Discussion Yes AND Pass)")
print(f"= {joint_count}/{total} = {joint_probability:.4f}")
print(f"= {joint_probability * 100:.2f}%")

print("\n2. Conditional Probability: P(Pass | Discussion Yes)")
print(
    f"= {joint_count}/{discussion_yes_count} "
    f"= {pass_given_discussion:.4f}"
)
print(f"= {pass_given_discussion * 100:.2f}%")

print("\n3. Conditional Probability: P(Pass | Discussion No)")
print(
    f"= {pass_without_discussion_count}/{discussion_no_count} "
    f"= {pass_given_no_discussion:.4f}"
)
print(f"= {pass_given_no_discussion * 100:.2f}%")
```

```python

import matplotlib.pyplot as plt
import numpy as np

# Expected Value and Variance
x = np.array([0, 1, 2, 3])
probabilities = np.array([0.0208, 0.1645, 0.4336, 0.3811])

expected_value = np.sum(x * probabilities)
variance = np.sum((x - expected_value)**2 * probabilities)
standard_deviation = np.sqrt(variance)

print("=" * 50)
print("EXPECTATION DECIDER - STATISTICAL SUMMARY")
print("=" * 50)
print(f"Expected Number of Students Passing: {expected_value:.2f}")
print(f"Variance: {variance:.4f}")
print(f"Standard Deviation: {standard_deviation:.4f}")

# Professional visualization
plt.figure(figsize=(10, 6))
bars = plt.bar(x, probabilities, width=0.6, edgecolor="black")

plt.plot(x, probabilities, marker="o", linewidth=2, markersize=7)

for bar, probability in zip(bars, probabilities):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height() + 0.012,
        f"{probability*100:.2f}%",
        ha="center",
        fontsize=10,
        fontweight="bold"
    )

plt.title("Expected Number of Students Passing", fontsize=16, fontweight="bold")
plt.xlabel("Number of Students Passing", fontsize=12)
plt.ylabel("Probability", fontsize=12)
plt.xticks(x)
plt.ylim(0, max(probabilities) + 0.10)
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```
